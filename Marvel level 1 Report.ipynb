## Task 6 : Battle-Test Your Model - Support Vector Machines

- After preprocessing and training an RBF SVM model, I obtained a baseline performance of 75.3% accuracy and 0.76 F1-score, showing moderate predictive strength in the wine dataset.
- I gradually injected Gaussian noise (0 → 1.0 std deviation) and retrained the model at each level. The performance declined steadily rather than collapsing abruptly.
- Accuracy dropped from 0.75 at no noise to a minimum of 0.669 at noise = 0.8, indicating roughly an 8% degradation. Interestingly, performance slightly recovered at noise = 1.0, reflecting the stochastic nature of Gaussian noise.
- Based on the sharpest performance dip, I identified 0.8 as the effective breakdown point, where the model showed maximum instability under corruption.
- Overall, the experiment demonstrated graceful degradation, suggesting that RBF-based SVM maintains reasonable robustness under moderate feature noise.

## Task 8 : KNN with Ablation Study

- A K-Nearest Neighbors (KNN) classifier was implemented on the Breast Cancer Wisconsin dataset to classify tumors as malignant (1) or benign (0). Irrelevant columns (id, Unnamed: 32) were removed, and features were normalized using StandardScaler to ensure proper distance calculation. An 80–20 stratified split was used with k=5 and Euclidean distance.
- The baseline model achieved approximately 95–97% accuracy, along with strong precision, recall, and F1-score. High recall indicates effective detection of malignant cases, which is critical in medical applications.
- A feature ablation study was performed by removing one feature at a time and retraining the model. Tumor size-related features (e.g., radius, perimeter, area) caused the most significant performance drop when removed, confirming their importance. Some features had minimal impact due to high correlation and redundancy.
- Overall, the results show that KNN performs well with proper scaling and that feature selection plays a key role in improving model robustness and interpretability.
